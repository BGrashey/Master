```python
import numpy as np
import matplotlib.pyplot as plt

from astropy.io import fits
from astropy.table import Table
from scipy.optimize import curve_fit


def wave_to_idx(lamda, wave_start=3470, wave_step=2):
    t = []
    for wl_min, wl_max in lamda:
        idx_start = int(np.round((wl_min - wave_start) / wave_step))
        idx_end = int(np.round((wl_max - wave_start) / wave_step)) + 1
        
        t.append((idx_start, idx_end))
    return t



bad_wl_ranges = [[4648,4686],[4854, 4868], [5140,5144], [5170, 5210]] # shows lots of spurious signal in both fields

bad_idxs = wave_to_idx(bad_wl_ranges)
bad_idxs



with fits.open("/data/hetdex/u/bgrashey/cubes/ssa22_fullfp_stack.fits") as hdul:
    data = hdul[0].data
    header = hdul[0].header
```

```python
import astropy.units as u
from astropy.coordinates import SkyCoord
from astropy.wcs import WCS
from astroquery.gaia import Gaia

# 1. Zentrum und Radius aus dem WCS deines Cubes bestimmen
wcs = WCS(header, naxis=2)  # naxis=2 ignoriert die spektrale Achse
ny, nx = data.shape[1], data.shape[2]

ra_center, dec_center = wcs.all_pix2world(nx / 2, ny / 2, 0)
center_coord = SkyCoord(
    ra=ra_center, dec=dec_center, unit=(u.deg, u.deg), frame="icrs"
)

# Radius des FoVs (z. B. Diagonale / 2)
corner_ra, corner_dec = wcs.all_pix2world(0, 0, 0)
fov_radius = (
    center_coord.separation(
        SkyCoord(ra=corner_ra, dec=corner_dec, unit=(u.deg, u.deg))
    )
    * 1.05
)

# 2. Gaia DR3 abfragen mit Filterkriterien
query = f"""
SELECT source_id, ra, dec, phot_g_mean_mag, bp_rp, ruwe
FROM gaiadr3.gaia_source
WHERE 1 = CONTAINS(
    POINT('ICRS', ra, dec),
    CIRCLE('ICRS', {center_coord.ra.deg}, {center_coord.dec.deg}, {fov_radius.deg})
)
AND phot_g_mean_mag BETWEEN 15.0 AND 19.5
AND ruwe < 1.2
"""

job = Gaia.launch_job_async(query)
stars_table = job.get_results()

# 3. Himmelskoordinaten (RA/Dec) in Pixelkoordinaten (x, y) umrechnen
star_coords = SkyCoord(
    ra=stars_table["ra"], dec=stars_table["dec"], unit=(u.deg, u.deg)
)
x_pix, y_pix = wcs.all_world2pix(star_coords.ra.deg, star_coords.dec.deg, 0)

stars_table["x"] = x_pix
stars_table["y"] = y_pix
```

```python
margin = 15
mask_inside = (
    (stars_table["x"] >= margin)
    & (stars_table["x"] < nx - margin)
    & (stars_table["y"] >= margin)
    & (stars_table["y"] < ny - margin)
    & (stars_table["phot_g_mean_mag"] < 16)
)
clean_stars = stars_table[mask_inside]
clean_stars.write("./clean_stars.fits", overwrite=True)
```

```python
clean_stars = Table.read("./clean_stars.fits")
```

```python
import numpy as np
from astropy.nddata import Cutout2D
from astropy.stats import sigma_clipped_stats

num_slabs = 5
bin_edges = np.linspace(0, data.shape[0], num_slabs + 1, dtype=int)

# Zentrale Indizes der Wellenlängenbänder (einmalig)
central_indices = [(bin_edges[i] + bin_edges[i + 1]) // 2 for i in range(num_slabs)]

clean_stars.sort("phot_g_mean_mag")

# Positionen der Quellen
N = len(clean_stars)  # Anzahl der Quellen
positions = [(clean_stars[i]["x"], clean_stars[i]["y"]) for i in range(N)]

# Jede Quelle bekommt ihre eigene Unterliste: [Quelle_0, Quelle_1, ...]
narrow_band_images = [[] for _ in range(N)]

# 1. Schleife über Slabs (Slab wird nur 1x kollabiert)
for i in range(num_slabs):
    idx_start = bin_edges[i]
    idx_end = bin_edges[i + 1]
    
    #slab_image , _, _ = sigma_clipped_stats(
    #    data[idx_start:idx_end, :, :], sigma=3.0, maxiters=5, axis=0
    #)
    
    # 2. Schleife über alle Quellen: Cutout in die jeweilige Quellen-Liste einsortieren
    for n in range(N):
        x0, y0 = int(round(positions[n][0])), int(round(positions[n][1]))
        half = 20  # a bit bigger than your 29x29 cutout, for safety
        sub_cube = data[idx_start:idx_end, y0-half:y0+half, x0-half:x0+half]
        slab_image = np.nanmedian(sub_cube, axis=0)
        
        local_pos = (positions[n][0] - x0 + half, positions[n][1] - y0 + half)


        cutout = Cutout2D(slab_image, position=local_pos, size=(29, 29))
        narrow_band_images[n].append(cutout.data)
```

```python
import numpy as np
from astropy.modeling import fitting, models

fitter = fitting.LevMarLSQFitter()
y, x = np.mgrid[:29, :29]

spaxel_scale = 0.5

crval3 = header['CRVAL3']
cdelt3 = header.get('CDELT3', header.get('CD3_3', 1.0)) 

nz = data.shape[0]  

wave = crval3 + np.arange(nz) * cdelt3

# Wellenlängen der Bin-Zentren (statt Pixel-Indizes)
central_wavelengths = wave[central_indices]

fwhm_per_star = []  # shape: (N, num_slabs)

for n in range(N):
    fwhm_n = []
    for i in range(num_slabs):
        img = narrow_band_images[n][i].copy()

        if np.isnan(img).any():
            img = np.nan_to_num(img, nan=np.nanmedian(img))

        bkg_guess = np.nanmedian(img)
        amp_guess = np.nanmax(img) - bkg_guess

        moffat = models.Moffat2D(
            amplitude=amp_guess,
            x_0=14.5,
            y_0=14.5,
            gamma=2.5,
            alpha=3.,
        )
        moffat.gamma.bounds = (0.8, 12.0)
        moffat.alpha.bounds = (2.9, 3.5)
        moffat.x_0.bounds = (10, 19)
        moffat.y_0.bounds = (10, 19)

        model_init = moffat + models.Const2D(amplitude=bkg_guess)
        fitted_model = fitter(model_init, x, y, img, maxiter=1000, filter_non_finite=True)

        alpha_fit = fitted_model[0].gamma.value
        beta_fit = fitted_model[0].alpha.value

        if fitter.fit_info["ierr"] not in [1, 2, 3, 4] or beta_fit <= 1.0:
            fwhm_n.append(np.nan)
            continue

        fwhm_pixels = 2 * alpha_fit * np.sqrt(2 ** (1 / beta_fit) - 1)
        fwhm_n.append(fwhm_pixels * spaxel_scale)

    fwhm_per_star.append(fwhm_n)

fwhm_per_star = np.array(fwhm_per_star)  # shape (N, num_slabs)

good_mask = np.nanmax(fwhm_per_star, axis=1) <= 10.
fwhm_per_star_clean = fwhm_per_star[good_mask]

fwhm_mean = np.nanmean(fwhm_per_star_clean, axis=0)
fwhm_std = np.nanstd(fwhm_per_star_clean, axis=0)

# Plot: einzelne Sterne (dünn/grau) + Mittelwert mit Fehlerbalken
for n in range(len(fwhm_per_star_clean)):
    plt.plot(central_wavelengths, fwhm_per_star_clean[n], color="gray", alpha=0.3)

plt.errorbar(central_wavelengths, fwhm_mean, yerr=fwhm_std, marker="o", color="C0", label="Mittelwert ± Std")
plt.xlabel("Wellenlänge [Å]")
plt.ylabel("FWHM [arcsec]")
plt.legend()
plt.show()
```
